```python
import os
from dotenv import load_dotenv
load_dotenv()
from langchain_openai import ChatOpenAI

from dotenv import load_dotenv

if os.environ['AICREDITS_API_KEY']:
    print("AICredits API Key is set.")
else:
    raise ValueError("AICredits API Key is not set.")
```

```python
llm_openai = ChatOpenAI(
    model="openai/gpt-5-nano",
    temperature=0.7,
    base_url="https://api.aicredits.in/v1",
    api_key=os.environ["AICREDITS_API_KEY"],
)
```

### **Guiding the Prompts**

```python
from langchain_core.prompts import PromptTemplate
result = llm_openai.invoke("Tell me a joke. Generate the output in key value pair format with the following keys: setup, punchline")
result.content
```

### **Using Pydantic Models**

```python
from pydantic import BaseModel, Field
class llm_schema(BaseModel):
    setup: str = Field(description="The setup for the joke")
    punchline: str = Field(description="The punchline for the joke")
```

```python
llm_structured_output = llm_openai.with_structured_output(llm_schema)

result = llm_structured_output.invoke("Tell me a joke")
print(result)
result.setup
```

### **Using TypedDict**

```python
from typing import TypedDict
class llm_schema_td(TypedDict):
    setup: str
    punchline:str
```

```python
llm_structured_output_typed_dict = llm_openai.with_structured_output(llm_schema_td)
result = llm_structured_output_typed_dict.invoke("Tell me a joke")
result
```
