### Setting up AICredits for all Models

```python
import os
from dotenv import load_dotenv
load_dotenv()
from langchain_openai import ChatOpenAI

from dotenv import load_dotenv

if os.environ['AICREDITS_API_KEY']:
    print("AICredits API Key is set.")
else:
    raise ValueError("AICredits API Key is not set.")
```

### LLM Call

```python
llm = ChatOpenAI(
    model="openai/gpt-5-nano",
    base_url="https://api.aicredits.in/v1",
    api_key=os.environ["AICREDITS_API_KEY"],
)
llm.invoke("Bro, tell me a joke").content
```

```python
import os
from langchain.chat_models import init_chat_model

llm_openai = init_chat_model(
    model="openai/gpt-5-nano",                # or your intended AICredits model name
    model_provider="openai",
    base_url="https://api.aicredits.in/v1",
    api_key=os.environ["AICREDITS_API_KEY"],
)
llm_openai.invoke("Hello, how are you?")
```
