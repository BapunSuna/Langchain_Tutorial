```python
import os
from dotenv import load_dotenv
load_dotenv()
from langchain_openai import ChatOpenAI

from dotenv import load_dotenv

if os.environ['AICREDITS_API_KEY']:
    print("AICredits API Key is set.")
else:
    raise ValueError("AICredits API Key is not set.")
```

```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from  langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm_openai = ChatOpenAI(model="gpt-5-mini",
                        base_url="https://api.aicredits.in/v1",
                        api_key=os.environ["AICREDITS_API_KEY"])
```

### **CHAIN WITH CUSTOM RUNNABLE**

```python
# TASK -1 [Prompt]

prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant"),
    ("human", "{input}")])
```

```python
# TASK - 2 [LLM]

llm_openai = ChatOpenAI(model="gpt-5-mini",
                        base_url="https://api.aicredits.in/v1",
                        api_key=os.environ["AICREDITS_API_KEY"])
```

```python
# TASK - 3 [Str Parser]

str_parser = StrOutputParser()
```

```python
# TASK - 4 [Custom Runnable]
from langchain_core.runnables import RunnableLambda

def dictionary_maker(text: str) -> dict:
    return {"text": text}

dictionary_maker_runnable = RunnableLambda(dictionary_maker)
```

```python
# TASK - 5 [Template for Post]
prompt_post = ChatPromptTemplate.from_messages(
    messages=[
        ("system", "You are a social media post generator."),
        ("human", "Create a post for the following text for LinkedIn {text}."),
    ]
)
```

```python
llm_openai = ChatOpenAI(model="gpt-5-mini",
                        base_url="https://api.aicredits.in/v1",
                        api_key=os.environ["AICREDITS_API_KEY"])
```

```python
str_parser = StrOutputParser()
```

```python
# CHAIN
chain = prompt_template | llm_openai | str_parser | dictionary_maker_runnable | prompt_post | llm_openai | str_parser
chain.invoke("What is the capital of India")
```
