### **CHAINS IN LANGCHAIN**

```python
import os
from dotenv import load_dotenv
load_dotenv()
from langchain_openai import ChatOpenAI

from dotenv import load_dotenv

if os.environ['AICREDITS_API_KEY']:
    print("AICredits API Key is set.")
else:
    raise ValueError("AICredits API Key is not set.")
```

```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm_openai = ChatOpenAI(model="gpt-5-mini",
                        base_url="https://api.aicredits.in/v1",
                        api_key=os.environ["AICREDITS_API_KEY"])
```

### **FIRST CHAIN**

```python
# TASK - 1 [PROMPT]
prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI Assistant."),
    ("human", "{input}"),
])
```

```python
# TASK - 2 [LLM]
llm_openai = ChatOpenAI(model="gpt-5-mini",
                        temperature=0,
                        base_url="https://api.aicredits.in/v1",
                        api_key=os.environ["AICREDITS_API_KEY"])
```

```python
# TASK - 3 [Str Parser]
str_parser = StrOutputParser()
```

### **CHAIN INVOCATION**

```python
chain = prompt_template | llm_openai | str_parser
chain.invoke("What is the capital of India")
```

```python
from langchain_core.runnables import RunnableSequence
chain_1 = RunnableSequence(prompt_template, llm_openai, str_parser)
chain_1.invoke("What is the capital of France?")
```
